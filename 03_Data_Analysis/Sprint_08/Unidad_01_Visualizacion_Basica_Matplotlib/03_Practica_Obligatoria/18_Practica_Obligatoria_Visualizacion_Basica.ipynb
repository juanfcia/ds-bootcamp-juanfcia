#### #Recreamos la primera figura del 1.2  

1. Las ciudades más pobladas, con los numeros de las poblaciones (tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento `ha` a "center"). (Colores: azul, gris, aceituna, marrón, con trasparencia)  

```python
# Ordenamos el dataFrame por población descendente y seleccionamos las 5 primeras filas
df_top_pob = df_cities.sort_values(by="population_total", ascending=False).head(5)

# Extraemos los nombres de las ciudades y sus poblaciones
ciudades = df_top_pob["city"].to_list()
frecuencias = df_top_pob["population_total"].to_list()

fig,ax = plt.subplots(figsize = (6,5))
ax.bar(ciudades, frecuencias, color = ["blue","grey","coral","olive", "brown"], alpha = 0.5)

ax.set_title("Ciudades más pobladas de California")
ax.set_xlabel("Ciudades")
ax.set_ylabel("Habitantes (miles)");
```

```python
# Bucle for para colocar la cifra de población en el centro superir de cada barra 
# Entenderlo y añadirlo al gráfico

# 3. Creamos el gráfico de barras. Guardamos el objeto 'bars' que devuelve ax.bar()
bars = ax.bar(ciudades, frecuencias, color = ["blue","grey","coral","olive", "brown"], alpha = 0.5)

# 4. AÑADIMOS LAS ANOTACIONES DE TEXTO (ETIQUETAS)

# Iteramos sobre cada barra individual en el objeto 'bars'
for bar in bars:
    height = bar.get_height()  # Obtenemos la altura de la barra (el valor de población)
    ax.text(
        bar.get_x() + bar.get_width()/2., # Posición X: El centro de la barra
        height + 20,                      # Posición Y: Un poco por encima de la barra (+20 unidades)
        f'{height:,}',                    # El texto a mostrar, formateado con comas (ej: 3,898)
        ha='center',                      # Alineación Horizontal: Centrado en la posición X
        va='bottom',                      # Alineación Vertical: Anclado en la parte inferior del texto
        fontsize=10                       # Tamaño de la fuente
    )
```

#### #Recreamos la segunda figura del 1.2  

2. Histograma del área total ocupada por las ciudades. (50 bins, fuente = "Bradley Hand ITC", tamaño 18, color azul acero o "steelblue")

```python
import matplotlib.patches as mpatches # Importamos la clase Patch

plt.hist(df_cities["area_total_sq_mi"], 
         bins = 50,
         alpha = 0.7,
         color = "steelblue"
         )

plt.title("Población por kilómetro cuadrado de agua", fontname = "Georgia") # Buscar cómo cargar la fuente pedida 
plt.xlabel("Superficie (km2)", fontname = "Georgia")
plt.ylabel("Número de Ciudades", fontname = "Georgia")

barra_simulada = mpatches.Patch(color = "steelblue",
                                alpha = 0.7,
                                label = "Hist Area"
                                )

plt.legend(handles = [barra_simulada], loc = "upper right");
```

#### #Recreamos la tercera figura del 1.2  

3. El scatter de área de agua de un ciudad con su poblacion.

```python
# Diagrama de dispersión de Población por kilómetro cuadrado de agua 

# Definimos un diagrama de dispersión con variables área de agua y población
# Además le asignamos un color verde, una transparencia de 0.5, title y dos labels por eje

plt.scatter(df_cities["area_water_km2"], df_cities["population_total"], c = "green", alpha = 0.5)
plt.title("Población por kilómetro cuadrado de agua")
plt.xlabel("Área de agua en la ciudad (km2)")
plt.ylabel("Población (miles habitantes)");


# Diccionario definir los estilos de la flecha
props_flecha = {
    "arrowstyle": "<->", # Aquí hay varios estilos de flecha definidos
    "connectionstyle": "arc3" # Aquí lo mismo con el tipo de conexión flecha y anotación
}

# Defino los ejes X e Y y sus valores máximo para las anotaciones de texto
x = df_cities["area_water_km2"]
y = df_cities["population_total"]
max_x_value = x.max()
max_y_value = y.max()

# Anotación para el outlier del eje x --> no consigo subir el texto en el eje Y
plt.annotate(
    "Outliers",
    xy = (max_x_value, df_cities.loc[x.idxmax(), "population_total"]), 
    xytext = (100, 3),                                                
    arrowprops = props_flecha,
    ha = "center"
)

# Anotación para el outlier del eje y --> no consigo subir el texto en el eje Y
plt.annotate(
    "Outliers",
    xy = (df_cities.loc[y.idxmax(), "area_water_km2"], max_y_value), 
    xytext = (100, 3),                                                
    arrowprops = props_flecha,
    ha = "center"
);
```

#### #Recreamos la cuarta figura del 1.2  

4. El boxplot de la altura de las ciudades.

```python
# Diagrama de caja de la altitud de las ciudades de California

plt.boxplot(
    df_cities["elevation_m"],
    vert = False,
    patch_artist=True, 
    medianprops={"color": "black"},
    boxprops={"facecolor": "red",
              "edgecolor": "black"}
) 

plt.xlabel("Altitud (m)")
plt.title("Diagrama de caja de la Altitud de las ciudades de California");
```
